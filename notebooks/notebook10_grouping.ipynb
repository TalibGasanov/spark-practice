```python
from pyspark.sql import SparkSession

DEFAULT_SPARK_REMOTE = "sc://localhost:15002"
spark = SparkSession.builder.remote(DEFAULT_SPARK_REMOTE).getOrCreate()
```

```python
spark.sql("use spark_catalog")
spark.sql("use schema northwind_db")
```

```python
df = spark.table("order_details")
display(df.limit(3))
```

```python
from pyspark.sql.functions import count, avg, round, expr, bround, sum, col

df_g = df.groupBy("OrderID").agg(
    count("ProductID").alias("ProductsNum"),
    round(avg("UnitPrice"),2).alias("AvgPrice"),
    bround(expr("SUM(UnitPrice * Quantity)")).alias("TotalAmount1"),
    sum(col("UnitPrice") * col("Quantity")).alias("TotalAmount2")
)

display(df_g)
```

```python
df_g = df.groupBy("OrderID").agg({
    "ProductID": "count",
    "UnitPrice": "avg"
})
display(df_g)
```
