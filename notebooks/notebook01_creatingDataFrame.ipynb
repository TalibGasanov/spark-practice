```python
from pyspark.sql import SparkSession

DEFAULT_SPARK_REMOTE = "sc://localhost:15002"
spark = SparkSession.builder.remote(DEFAULT_SPARK_REMOTE).getOrCreate()

spark.version
```

```python
from pyspark.sql import Row
from pyspark.sql.functions import col, concat, lit


row01 = Row("Ivan", "Ivanov")
row02 = Row("Petr", "Petrov")

schema = ["FirstName", "SecondName"]
df = spark.createDataFrame([row01, row02], schema)
df = df.withColumn("FullName", concat("FirstName", lit(" "), "FirstName"))
display(df)
```

```python
from pyspark.sql.functions import expr, col

df2 = df.select(expr("FirstName as Name"), "SecondName")
display(df2)
```

```python
from pyspark.sql.functions import expr, col

df2 = df.selectExpr("FirstName as Name", "SecondName", "concat(FirstName, ' ', SecondName) as FullName")
display(df2)
```

```python
from pyspark.sql.functions import lit, expr

df3 = df.select(expr("FirstName as Name"), (lit(1000) * 5).alias("Salary"), expr("2*2 as 2x2"))
display(df3)
```

```python
df4 = (
        df.withColumnRenamed("FirstName", "First name")
        .withColumnRenamed("SecondName", "Second name")
        .withColumnRenamed("FullName", "Full Name")
        .withColumn("Age", lit("25")))

display(df4)

df4 = df4.withColumn("Age", col("Age").cast("Integer"))

display(df4)

count = df4.select("Age").distinct().count()
print(count)
```

Если передать простой список, то это будет одна колонка

```python
lst = [1,2,3]
df = spark.createDataFrame(lst, schema=["id"])
df.printSchema()
display(df)
```

Если передать список списков, то каждый список из списка становится строкой

```python
lst = [
    [1,"Иванов"],
    [2,"Петров"],
    [3,"Сидоров"]
]
df = spark.createDataFrame(lst, schema = "id Integer, name String")
df.printSchema()
display(df)
```

```python
lst = [
    [1, 2, 3],
    ["Иванов", "Петров", "Сидоров"],
]
df = spark.createDataFrame(lst, ["a","b","c"])
df.printSchema()
display(df)
```

```python
squares = [ x ** 2 for x in range(3)]
df2 = spark.createDataFrame(squares, "square Integer")

print(squares)
df2.printSchema()
display(df2)

df2 = spark.createDataFrame(squares, ["square"])
df2.printSchema()
display(df2)
```
